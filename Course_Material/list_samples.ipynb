## 1. Building lists with range
`range()` is perfect for generating consecutive integers. Wrap it in `list()` when you need an actual list.

```python
list1 = list(range(1, 9))list1# Extra: start at 0 and step by 2evens_under_20 = list(range(0, 20, 2))evens_under_20
```

## 2. Sorting in reverse order
Use `sort(reverse=True)` to quickly flip order in place. Ideal for ranking scores from greatest to smallest.

```python
squares = [1, 4, 9, 16, 25, 36, 49, 64]squares.sort(reverse=True)squares# Extra: sort a copy without touching the originalsorted_copy = sorted([3, 10, 2, 7], reverse=True)sorted_copy
```

## 3. Mutating lists (append/remove/pop/insert)
Mutating methods help maintain running state. Remember that they all operate in place.

```python
squares = [1, 4, 9, 16, 25, 36, 49, 64]squares.append(9 ** 2)squares.remove(1)last = squares.pop()  # remove last element and capture itsquares.insert(3, 9_999)squares, last# Extra: replace the element at index 2squares[2] = 123squares
```

## 4. Size and membership checks
`len()` plus `in`/`not in` keep your guards simple. Useful for bounds checking or duplicate detection.

```python
len(squares)25 in squares999 in squares# Extra: find min/max after defending against emptiesmin(squares), max(squares)
```

## 5. Nested list access
Index into nested lists one level at a time. This pattern mirrors matrix access in many CCC problems.

```python
list2 = [[1, 2, 3, 4], [2, 3, 4, 5], 3, 8]list2[1][2]# Extra: flatten the numeric rowsflattened = list2[0][:] + list2[1][:]flattened
```

## 6. List comprehensions
Comprehensions condense loops into one line—great for generating or filtering values.

```python
squares1 = [x ** 2 for x in range(1, 9)]squares1# Extra: keep only squares greater than 20big_squares = [x for x in squares1 if x > 20]big_squares
```

## 8. Practice challenge
Prompt the user for a list of scores (space-separated), drop the lowest score, and print the average of the rest.

```python
raw_scores = input("Enter scores separated by spaces: ")scores = [int(x) for x in raw_scores.split()]if len(scores) <= 1:    print("Need at least two scores.")else:    trimmed = scores[:]    trimmed.remove(min(trimmed))    average = sum(trimmed) / len(trimmed)    print(f"Trimmed scores: {trimmed}")    print(f"Average: {average:.2f}")
```
