# Week 11 – List Algorithms

**Goals**: sorting, searching, merging, sublists, efficiency.

## Key Concepts
- Bubble sort vs built-in `sorted()`
- Linear vs Binary search
- Merging two sorted lists

```python
# Bubble sort
nums = [5,3,8,4,2]
n = len(nums)
for i in range(n-1):
    for j in range(n-i-1):
        if nums[j] > nums[j+1]:
            nums[j], nums[j+1] = nums[j+1], nums[j]
nums
# Linear search
nums = [10,20,30,40]; target = 30
idx = -1
for i,x in enumerate(nums):
    if x == target:
        idx = i; break
idx
# Binary search (requires sorted list)
nums = [1,3,5,7,9,11,13]; target = 9
lo, hi = 0, len(nums)-1; found = None
while lo <= hi:
    mid = (lo+hi)//2
    if nums[mid] == target:
        found = mid; break
    elif nums[mid] < target:
        lo = mid+1
    else:
        hi = mid-1
found
# Merge two sorted lists
a = [1,3,5]; b = [2,4,6]
i=j=0; merged=[]
while i < len(a) and j < len(b):
    if a[i] < b[j]:
        merged.append(a[i]); i+=1
    else:
        merged.append(b[j]); j+=1
merged.extend(a[i:]); merged.extend(b[j:])
merged
```

## Practice 2 – Remove Duplicates (keep order)

```python
nums = [1,2,2,3,4,4,5]
seen=set(); unique=[]
for x in nums:
    if x not in seen:
        unique.append(x); seen.add(x)
unique
```
