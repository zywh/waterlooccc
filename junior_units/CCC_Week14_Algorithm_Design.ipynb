## Brute Force vs Greedy

```python
# Brute force O(n^2) vs O(n) tracking min/max
nums = [3,10,6,1,8]
best = 0
for i in range(len(nums)):
    for j in range(i+1, len(nums)):
        best = max(best, abs(nums[i]-nums[j]))
best
mn = mx = nums[0]
for x in nums[1:]:
    mn = min(mn, x); mx = max(mx, x)
mx - mn
```

## Practice – Coin Change (Greedy U.S. coins)

```python
N = 87
coins = [25,10,5,1]
cnt=0
for c in coins:
    take = N//c
    cnt += take; N -= take*c
cnt
```

## Practice – Interval Scheduling by Finish Time

```python
intervals = [(1,4),(3,5),(0,6),(5,7),(8,9),(5,9)]
intervals.sort(key=lambda x: x[1])
ans=[]; last_end=-1
for s,e in intervals:
    if s >= last_end:
        ans.append((s,e)); last_end = e
ans
```

## Homework – Task Packing (Greedy by duration)

```python
D = 10
tasks = [2,3,1,2,5,4]
tasks.sort(); time=taken=0
for t in tasks:
    if time + t <= D:
        time += t; taken += 1
taken
```
