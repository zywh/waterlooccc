## Key Concepts
- `split()`, `join()`
- `find()`, `in`, `count()`
- `replace(old, new)`
- Immutability: string ops return new strings

```python
# Example: split and join
text = "apple,banana,cherry"
items = text.split(",")
print(items)
" | ".join(items)
# Example: count occurrences
sentence = "hello world hello python"
sentence.count("hello")
# Example: palindrome check
word = input("Enter a word: ").lower().replace(" ", "")
print("Palindrome!" if word == word[::-1] else "Not a palindrome.")
```

## Practice 1 – Reverse Words

```python
sentence = input("Enter a sentence: ")
words = sentence.split()
" ".join(reversed(words))
```

## Practice 2 – Longest Word

```python
sentence = input("Enter a sentence: ")
words = sentence.split()
max(words, key=len)
```

## Practice 3 – Anagram Check

```python
a = input("First string: ").replace(" ", "").lower()
b = input("Second string: ").replace(" ", "").lower()
print("Anagram!" if sorted(a) == sorted(b) else "Not an anagram.")
```

## Example – Caesar Cipher Decode (shift -3)

```python
text = input("Enter encoded text: ").upper()
decoded = []
for ch in text:
    if ch.isalpha():
        decoded.append(chr((ord(ch) - 65 - 3) % 26 + 65))
    else:
        decoded.append(ch)
"".join(decoded)
```

## Homework
Write a program that outputs each unique word and its frequency, sorted alphabetically.

```python
text = input("Enter text: ").lower()
freq = {}
for w in text.split():
    freq[w] = freq.get(w, 0) + 1
for w in sorted(freq):
    print(w, ":", freq[w])
```
