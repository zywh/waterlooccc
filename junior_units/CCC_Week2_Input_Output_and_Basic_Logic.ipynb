I Learning Objectives
• Learn to take user input using `input()`.

I Key Concepts
1. User Input: Use input() to read from keyboard (always returns a string).
2. Type Conversion: Wrap input with int() or float() when you need numbers.
3. Conditionals: Use if/elif/else to make decisions based on conditions.
4. Boolean Expressions: Evaluate to True or False.

## Example 1 – Reading User Input

# Ask for name and age

```python
name = input("What is your name? ").strip()
age_text = input("How old are you? ").strip()
try:
    age = int(age_text)
except ValueError:
    print("Please enter your age using digits only.")
else:
    greeting_name = name or "friend"
    print(f"Hello {greeting_name}, next year you will be {age + 1} years old!")
```

## Example 2 – Using if/else Statements

```python
temperature_text = input("Enter the temperature in °C: ").strip()
try:
    temperature = float(temperature_text)
except ValueError:
    print("Temperature must be numeric, like 18 or -2.5.")
else:
    if temperature > 30:
        print("It's a hot day!")
    elif 20 <= temperature <= 30:
        print("It's a nice day.")
    elif 10 <= temperature < 20:
        print("Grab a light jacket.")
    else:
        print("It's a bit cold today.")
```

## Example 3 – Reading Multiple Values From One Line
Many CCC problems supply several numbers on a single line. We can call `split()` to break the text into pieces before converting each one.

```python
raw_scores = input("Enter three quiz scores separated by spaces: ").strip()
try:
    values = [float(item) for item in raw_scores.split()]
except ValueError:
    print("Only numbers are allowed.")
else:
    if len(values) != 3:
        print("Please enter exactly three numbers.")
    else:
        average = sum(values) / len(values)
        print(f"Average score: {average:.1f}")
        if average >= 80:
            print("Great job! You're above the honour roll threshold.")
        elif average >= 50:
            print("You passed. Keep practising to boost your average.")
        else:
            print("Let's review the material together.")
```

### Practice 1 – Even or Odd Checker

Ask the user to enter a number and print whether it’s even or odd.
Hint: Use the modulus operator %.
Solution:

```python
num_text = input("Enter a whole number: ").strip()
if num_text.lstrip("-").isdigit():
    num = int(num_text)
    if num % 2 == 0:
        print(f"{num} is even.")
    else:
        print(f"{num} is odd.")
else:
    print("That wasn't a valid integer. Try again with digits only.")
```

### Practice 2 – Age Group Classifier

Ask the user for their age. Print 'Child' if under 13, 'Teen' if between 13–19, else 'Adult'.
Solution:

```python
age_text = input("Enter your age: ").strip()
if not age_text.isdigit():
    print("Age must be a positive whole number.")
else:
    age = int(age_text)
    if age < 13:
        print("Child")
    elif age <= 19:
        print("Teen")
    else:
        print("Adult")
```

### Practice 3 – Pass/Fail Checker

Ask for a student’s score. Print 'Pass' if score ≥ 50, otherwise 'Fail'.
Solution:

```python
score_text = input("Enter your score (0-100): ").strip()
try:
    score = float(score_text)
except ValueError:
    print("Score must be numeric.")
else:
    if not 0 <= score <= 100:
        print("Score must be between 0 and 100.")
    elif score >= 50:
        print("Pass")
    else:
        print("Fail")
```

I Boolean Logic Examples

```python
x = 10
y = 5
z = -3
print(f"x > y? {x > y}")
print(f"x == y? {x == y}")
print(f"x != y? {x != y}")
print(f"z < 0? {z < 0}")
if x > 0 and y > 0:
    print("Both x and y are positive.")
if x == 10 or y == 10:
    print("At least one variable equals 10.")
if not (z > 0):
    print("z is not positive.")
```

I Homework Challenge
Ask the user for a number of hours worked and hourly wage, then calculate total pay. If hours > 40, pay
1.5 × hourly rate for the extra hours.
Expected Solution:

```python
hours_text = input("Enter hours worked: ").strip()
rate_text = input("Enter hourly wage: ").strip()
try:
    hours = float(hours_text)
    rate = float(rate_text)
except ValueError:
    print("Please enter numeric values for hours and wage.")
else:
    if hours > 40:
        overtime_hours = hours - 40
        pay = 40 * rate + overtime_hours * rate * 1.5
    else:
        pay = hours * rate
    print(f"Total pay: ${pay:.2f}")
```
