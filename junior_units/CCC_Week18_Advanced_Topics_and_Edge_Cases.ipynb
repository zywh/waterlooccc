## Recursion – Tree Sum

```python
tree = {1:[2,3],2:[4],3:[],4:[]}
values = {1:5,2:3,3:2,4:4}
def tree_sum(u):
    return values[u] + sum(tree_sum(v) for v in tree[u])
tree_sum(1)
```

## Edge Cases – Safe Average & Normalize

```python
def safe_average(nums): return 0 if not nums else sum(nums)/len(nums)
def normalize(s): return " ".join(s.strip().split())
safe_average([]), safe_average([2,4,6]), normalize("  hello   world   ")
```

## Practice

```python
def dfs_depths(tree, root=1):
    depth={}
    def dfs(u,d):
        depth[u]=d
        for v in tree.get(u,[]): dfs(v,d+1)
    dfs(root,0); return depth
dfs_depths({1:[2,3],2:[4,5],3:[],4:[],5:[]})
```
