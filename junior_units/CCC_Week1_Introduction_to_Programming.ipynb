## Example 1: Hello World and Simple Math

I More Examples and Tips
You can combine strings and variables using commas or formatted strings (f-strings).

```python
name = "Alice"
age = 15
print("Name:", name, "| Age:", age)
# Using f-string for better formatting
print(f"My name is {name} and I am {age} years old.")
```
