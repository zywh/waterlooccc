## Example 1 – Simple Function

```python
def greet(name):
    return f"Hello, {name}!"
print(greet("Alice"))
```

## Example 2 – Compute Average

```python
def average(nums):
    if not nums:
        return 0.0
    total = 0
    for x in nums:
        total += x
        return total / len(nums)
print(average([5, 8, 10]))
```

### Practice 1 – Max of Three

Write function max3(a,b,c) returning the maximum of three numbers.
Solution:

```python
def max3(a, b, c):
    m = a
    if b > m:
        m = b
    if c > m:
        m = c
    return m
print(max3(3, 9, 5))
```

### Practice 2 – Count Letters

Write count_letter(s, ch) that returns how many times ch appears in s.
Solution:

```python
def count_letter(s, ch):
    cnt = 0
    for c in s:
        if c == ch:
            cnt += 1
    return cnt
print(count_letter("banana", "a"))
```

### Practice 3 – Validate Password

Write is_valid(pw) that returns True if length ≥ 8 and includes a digit and a letter.
Solution:

```python
def is_valid(pw):
    has_digit = False
    has_alpha = False
    if len(pw) < 8:
        return False
    for ch in pw:
        if ch.isdigit():
            has_digit = True
        if ch.isalpha():
            has_alpha = True
    return has_digit and has_alpha
print(is_valid("abc12345"))
```
