**Goals:** Nested/combined conditionals, `break`/`continue`, sentinel-controlled loops, and small simulations.

## Key Concepts
- **Nested if** for multi-level logic
- **Logical operators**: `and`, `or`, `not`
- **Loop control**: `break` (exit loop), `continue` (skip to next iteration)
- **Sentinel input** loops (e.g., until user types `stop`/`0`)

## Example 1 – Nested If Statements

```python
x = 7  # try other numbers
if x > 0:
    if x % 2 == 0:
        print("Positive even number")
    else:
        print("Positive odd number")
elif x < 0:
    print("Negative number")
else:
    print("Zero")
```

## Example 2 – While Loop with a Condition

```python
count = 0
while count < 5:
    print("Count =", count)
    count += 1
```

## Example 3 – Using break and continue

```python
for i in range(1, 10):
    if i == 5:
        continue   # skip 5
    if i == 8:
        break      # stop at 8
    print(i)
```

## Practice 3 – Multiplication Table (Skip a Row)
Print the 1–10 multiplication tables, but **skip** the table for number 5.

```python
tables = []
for i in range(1, 11):
    if i == 5:
        continue
    row = [f"{i}x{j}={i*j}" for j in range(1, 11)]
    tables.append(row)
tables  # inspect result
```

## Challenge – Number Classification
Ask for 10 numbers. Count how many are even, odd, divisible by 3, and divisible by 5.

```python
# Example (non-interactive demo). Replace 'data' with user input in class.
data = [3, 10, 12, 7, 25, 30, 1, 2, 5, 18]
even = odd = div3 = div5 = 0
for n in data:
    if n % 2 == 0: even += 1
    else: odd += 1
    if n % 3 == 0: div3 += 1
    if n % 5 == 0: div5 += 1
(even, odd, div3, div5)
```
