## Faster Lookups

```python
arr = list(range(10000)); targets = [3,7000,9999]
hits = sum(1 for t in targets if t in arr)           # baseline
s = set(arr)
hits_fast = sum(1 for t in targets if t in s)        # using set
(hits, hits_fast)
```

## Pair Sum in O(n)

```python
arr = [2,7,11,15]; T = 9
seen=set(); ans=None
for x in arr:
    if T - x in seen: ans=(T-x,x); break
    seen.add(x)
ans
```

## Prefix Sums

```python
nums = [1,2,3,4,5]
pref=[0]
for x in nums: pref.append(pref[-1]+x)
l,r = 1,3
pref[r+1]-pref[l]
```
