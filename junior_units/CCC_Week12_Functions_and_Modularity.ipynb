## Key Concepts
- `def` to define functions
- Parameters & return values
- Recursion with base case
- Modular design

```python
def greet(name):
    print("Hello,", name)
greet("Alice")
```

```python
def add(a,b): return a+b
add(5,3)
```

```python
def square(x): return x*x
def cube(x): return x*square(x)
cube(3)
```

```python
# Recursion: factorial
def factorial(n):
    return 1 if n<=1 else n*factorial(n-1)
factorial(5)
```

```python
# Recursion: fibonacci (simple)
def fib(n):
    return n if n<=1 else fib(n-1)+fib(n-2)
[fib(i) for i in range(6)]
```

## Practice 1 – Average Calculator

```python
def average(nums): return sum(nums)/len(nums)
average([5,10,15,20])
```

## Practice 2 – Count Vowels

```python
def count_vowels(text):
    return sum(ch in "aeiou" for ch in text.lower())
count_vowels("Function")
```

## Practice 3 – Sum of Digits (Recursive)

```python
def sum_digits(n): return 0 if n==0 else n%10 + sum_digits(n//10)
sum_digits(12345)
```

## Homework
Write modular program: read numbers → compute median → print result.

```python
def read_numbers():
    return list(map(int, input("Enter numbers: ").split()))
def median(nums):
    nums = sorted(nums)
    n=len(nums); mid=n//2
    return (nums[mid-1]+nums[mid])/2 if n%2==0 else nums[mid]
def main():
    data = read_numbers()
    print("Median =", median(data))
# main()  # Uncomment to run interactively
```
