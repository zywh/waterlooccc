# Week 3 – Loops Basics

I Learning Objectives
• Understand repetition and why loops are useful in programming.
• Learn syntax for while loops and for loops.
• Avoid infinite loops by setting proper conditions.
• Use loops to sum, count, and iterate over numbers or lists.
I Key Concepts
1. While Loop: Repeats while a condition is True.
2. For Loop: Repeats for a specific number of times using range().
3. Loop Control: Use break to exit and continue to skip to next iteration.
4. Counters and Accumulators: Track running totals and counts inside loops.

## Example 1 – While Loop Counting

```python
count = 1
while count <= 5:
    print("Count =", count)
    count += 1
print("Loop finished!")
```

## Example 2 – For Loop with range()

```python
for i in range(1, 6):
    print("Number:", i)
# range(start, stop) → stops before stop value
```

## Example 3 – Summing Numbers 1 to N

```python
n = int(input("Enter a number: "))
total = 0
for i in range(1, n + 1):
    total += i
print("Sum =", total)
```

I Practice Problems

### Practice 1 – Countdown Timer

Write a program that counts down from 10 to 1 and then prints 'Blast off!'.
Hint: Use range(10, 0, -1).
Solution:

```python
for i in range(10, 0, -1):
    print(i)
print("Blast off!")
```

### Practice 2 – Sum of Even Numbers

Compute the sum of all even numbers between 1 and 100.
Solution:

```python
total = 0
for i in range(2, 101, 2):
    total += i
print("Sum of even numbers:", total)
```

### Practice 3 – Find Factorial

Ask user for n and compute n! using a loop.
Solution:

```python
n = int(input("Enter a number: "))
fact = 1
for i in range(1, n + 1):
    fact *= i
print(f"The factorial of {n} is {fact}")
```

I Loop Control Statements

# Break Example

```python
# Break example: stop the loop when i == 5
for i in range(1, 10):
    if i == 5:
        print(f"Stopped at {i}")
        break
    print(i)

print("Loop ended (completed or broken).")
```

continue

```python
# Continue example: skip numbers divisible by 3
for i in range(1, 11):
    if i % 3 == 0:
        print(f"Skipping {i}")
        continue
    print(f"Processed {i}")
```
