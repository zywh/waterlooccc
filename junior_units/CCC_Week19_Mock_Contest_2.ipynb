## Problem A – Word Compression (J2)

```python
def compress(s):
    if not s: return ""
    out=[]; cur=s[0]; cnt=1
    for ch in s[1:]:
        if ch==cur: cnt+=1
        else: out.append(cur+str(cnt)); cur, cnt = ch,1
    out.append(cur+str(cnt))
    return "".join(out)
compress("aaabbbcca")
```

## Problem B – Meeting Rooms (Greedy Sweeping)

```python
intervals = [(0,30),(5,10),(15,20)]
events = []
for s,e in intervals:
    events.append((s,1)); events.append((e,-1))
events.sort()
cur = ans = 0
for _,d in events:
    cur += d; ans = max(ans, cur)
ans
```

## Problem C – Shortest Path in Grid (BFS)

```python
from collections import deque
grid = [
    ['.','.','#','.'],
    ['#','.','.','.'],
    ['.','#','.','.'],
    ['.','.','.','.']
]
R,C = len(grid), len(grid[0])
dist = [[-1]*C for _ in range(R)]
q = deque([(0,0)]); dist[0][0]=0
dirs=[(-1,0),(1,0),(0,-1),(0,1)]
while q:
    r,c = q.popleft()
    for dr,dc in dirs:
        nr,nc = r+dr,c+dc
        if 0<=nr<R and 0<=nc<C and grid[nr][nc] != '#' and dist[nr][nc] == -1:
            dist[nr][nc] = dist[r][c] + 1; q.append((nr,nc))
dist[R-1][C-1]
```
