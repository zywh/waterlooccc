# String Operations Workbook
Explore the string manipulation patterns used in `template/string.py`. Each section explains the original sample, demonstrates it, and adds an extra example you can adapt for CCC problems.

## 1. Base string setup
We'll reuse the same `ss` variable from the template so the behaviors remain familiar.

```python
ss = "sammy shark"
ss
```

## 2. Changing case
The original sample uppercases and capitalizes the string. Use casing helpers when normalizing user input or formatting messages.

```python
ss.upper(), ss.capitalize()
# Extra: title case for headlines
ss.title()
```

## 3. Slicing and stepping
Slices pull substrings using `[start:stop:step]`. Negative steps walk backward.

```python
ss[2:8:2]  # every other letter starting at index 2
ss[::-1]   # reverse the whole string
# Extra: grab the last four characters
ss[-4:]
```

## 4. Lengths and counts
Counting characters helps with constraints like “How many vowels?” or “Does the password have enough characters?”

```python
len(ss), ss.count("a")
# Extra: count any vowel using a comprehension
sum(ss.count(v) for v in "aeiou")
```

## 5. Character property checks
These boolean helpers quickly validate input. Combine with conditionals to branch logic.

```python
ss.isnumeric(), ss.isupper(), ss.islower(), ss.startswith("sa"), ss.endswith("victor")
# Extra: check if the stripped string is alphabetic
ss.replace(" ", "").isalpha()
```

## 6. Counting emoji markers
The template counts happy/sad emoticons by splitting. Use this approach when markers do not overlap.

```python
status_log = "I'm happy-:).I'm sad-:(  I'm happy again-:)"
happy = len(status_log.split("-:)")) - 1
sad = len(status_log.split("-:(")) - 1
happy, sad
# Extra: clean double spaces before recounting
cleaned_log = status_log.replace("  ", " ").strip()
cleaned_log.count("-:)"), cleaned_log.count("-:(")
```

## 7. Splitting into lists and converting
`split()` builds lists of substrings. Combine with comprehensions to coerce types.

```python
s_list = "1 2 5 6 9 10"
tokens = s_list.split()[2:6]
nums = [int(x) for x in tokens]
tokens, nums
# Extra: compute the sum in one line
sum(int(x) for x in s_list.split())
```

## 8. Removing characters
Replace or split-join removes unwanted symbols (e.g., spaces, punctuation).

```python
s3 = "abcaaabcdddaaa"
''.join(s3.split('a')), s3.replace('a', '')
# Extra: remove multiple letters by chaining replace
s3.replace('a', '').replace('d', '')
```

## 9. Joining lists
`join()` glues lists into one string. Useful for CSV output or readable summaries.

```python
items = ['sharks', 'crustaceans', 'plankton']
', '.join(items)
# Extra: present the list as a human sentence
if len(items) == 1:
    human = items[0]
else:
    human = f"{', '.join(items[:-1])} and {items[-1]}"
human
```

## 10. Formatting output
The template uses percent-formatting; here we show both percent-style and f-strings for timestamps.

```python
year, month, day, hour, minute = 2017, 1, 23, 1, 4
"%d-%02d-%02d  %02d:%02d" % (year, month, day, hour, minute)
# Extra: equivalent f-string with timezone label
f"{year}-{month:02d}-{day:02d} {hour:02d}:{minute:02d} UTC"
```

## 11. Advanced f-string patterns
F-strings format data inline: embed expressions, control width/precision, and apply alignment without leaving the literal.

```python
team = "Waterloo CCC"
points = 1234.5678
rank = 5
f"{team} earned {points:.2f} points and sits #{rank}"
# Expressions: compute inline and pad with leading zeros
runs = 7
target = 10
f"Short by {target - runs:02d} problems"
# Alignment with width, great for scoreboards
for place, name, score in [(1, "Alice", 92), (2, "Bob", 88), (3, "Chen", 88)]:
    print(f"{place:>2}. {name:<10} | {score:>3}")
```

## 12. Practice challenge
Combine what you learned: normalize a phrase, remove punctuation, and count vowels.

```python
phrase = input("Enter a phrase: " )
normalized = ''.join(ch for ch in phrase.lower() if ch.isalpha())
vowel_total = sum(normalized.count(v) for v in "aeiou")
print(f"Normalized: {normalized}")
print(f"Vowel count: {vowel_total}")
```
